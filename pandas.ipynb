```python
# NOTE: MERGE TWO DATAFRAME

df1=pd.DataFrame({
    'id':[1,2,3,4],
    'name':['Aman','Ram','Rahul','Geeta']
})

df2=pd.DataFrame({
    'id':[3,4,5,6],
    'course':['GenAI','Python','WEB-Dev','ML']
})

yes=pd.merge(df1,df2,'inner','id') #(left,right,how,on)
yes.head()


```

```python
yes=pd.merge(df1,df2,'left','id') #(left,right,how,on)
yes.head()
```

```python
yes=pd.merge(df1,df2,'right','id') #(left,right,how,on)
yes.head()
```

```python
yes=pd.merge(df1,df2,'outer','id') #(left,right,how,on)
yes.head()
```

```python
# NOTE: if we have given two labels which are contain same value but diffrent label name so how we perform such as
df3=pd.DataFrame({
    'id1':[1,2,3,4],
    'name':['Aman','Ram','Rahul','Geeta']
})

df4=pd.DataFrame({
    'id2':[3,4,5,6],
    'course':['GenAI','Python','WEB-Dev','ML']
})

yes=pd.merge(df3,df4,'inner',left_on='id1',right_on='id2')
yes.head()
```

```python
# for changing position 
woo = yes[['id1','id2','name','course']]
woo.head()
```

```python
# sometimes we have to join the dataframe or series not on the basis of some id or primary key sometimes we have to concatenate basis of row or columns
```

```python
df1=pd.DataFrame({
    'id':[1,2,3,4],
    'name':['Aman','Ram','Rahul','Geeta']
})

df2=pd.DataFrame({
    'id':[3,4,5,6],
    'course':['GenAI','Python','WEB-Dev','ML']
})

waah=pd.concat([df1,df2],axis=0,ignore_index=True) # on the basis of row beacuse axis=0
waah                # only print variable not head 
```

```python
waah2=pd.concat([df1,df2],axis=1) # on the basis of columns beacuse axis=1
waah2  
```

```python
# for diffrenciate the two or more rows 

fun=pd.concat([df1,df2],axis=0,keys=['df1','df2'])
fun
```
