```python
df['Brand'] = df['Brand'].fillna('Unknown')
df['Category'] = df['Category'].fillna('Others')
df['SubCategory'] = df['SubCategory'].fillna('Others')
df['Description'] = df['Description'].fillna('Not Available')
df['BreadCrumbs'] = df['BreadCrumbs'].fillna('Not Available')
```

```python
df
```

```python
df.isnull().sum()
```

```python
df['Brand'] = df['Brand'].fillna('Unknown')
df['Price'] = df['Price'].fillna('Others')
df['DiscountedPrice'] = df['DiscountedPrice'].fillna('None')
df['Quantity'] = df['Quantity'].fillna('Not Available')
df['BreadCrumbs'] = df['BreadCrumbs'].fillna('Not Available')
```

```python
df.isnull().sum()
```

```python
df['Name'] = df['Name'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
df['DiscountedPrice'] = pd.to_numeric(df['DiscountedPrice'], errors='coerce')
```

```python
df
```

```python
df.head(10)
```

```python
df['Quantity_Clean'] = df['Quantity'].str.extract(r'(\d+)')
df['Quantity_Clean'] = pd.to_numeric(df['Quantity_Clean'], errors='coerce')
```

```python
df.head()
```

```python
df['DiscountPercent'] = ((df['Price'] - df['DiscountedPrice']) / df['Price']) * 100
```
